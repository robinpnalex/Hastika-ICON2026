# Task B -- confirm one-layer reinitialization

Run 5 found that resetting only MuRIL's final encoder layer scored 0.6102 OOF macro-F1 versus 0.6013 for the established two-layer reset, using model seed 42. This notebook checks whether that advantage survives model seeds 43 and 44.

For each seed, both variants use the same shared TAPT checkpoint, deduplicated Task B data, fixed five-fold split (split seed 42), six epochs, mean+max pooling, FGM, EMA, class weighting, no auxiliary head, and `--select last`. The only intended model difference is `--reinit-layers 2` versus `--reinit-layers 1`.

Expected runtime: about 6--7 hours on a T4, including one TAPT pass. Upload this notebook to Kaggle with a GPU and Internet enabled, then use **Save Version -> Save & Run All**. This is a local confirmation experiment, not a submission notebook.

```python
import os, pathlib, shutil, subprocess, sys

WORK = "/kaggle/working/hastika"
if os.path.isdir(WORK + "/.git"):
    subprocess.run(["git", "-C", WORK, "pull", "--ff-only"], check=True)
else:
    subprocess.run(["git", "clone", "-q", "-b", "task-b", "--depth", "1",
                    "https://github.com/robinpnalex/Hastika-ICON2026.git", WORK], check=True)
os.chdir(WORK)
os.environ["PYTHONPATH"] = os.path.join(WORK, "src")
sys.path.insert(0, os.path.join(WORK, "src"))
pathlib.Path("artifacts/logs").mkdir(parents=True, exist_ok=True)
print("repo:", os.getcwd())
subprocess.run(["git", "log", "-1", "--oneline"], check=True)
subprocess.run('pip install -q emoji ftfy sentencepiece protobuf "transformers>=4.45,<6"',
               shell=True, check=True)
import torch
assert torch.cuda.is_available(), "no GPU -- set Accelerator in the sidebar"
print("gpu:", torch.cuda.get_device_name(0))

def run(cmd, log=None):
    print("$", " ".join(cmd), flush=True)
    fh = open(log, "w") if log else None
    p = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                         text=True, bufsize=1)
    for line in p.stdout:
        sys.stdout.write(line)
        if fh:
            fh.write(line)
    p.wait()
    if fh:
        fh.close()
    if p.returncode:
        raise RuntimeError(f"exit {p.returncode}: {cmd}")
```

## 1. Build one shared TAPT checkpoint

TAPT is run once and reused by all four classifier runs. Keeping this checkpoint shared prevents the confirmation from mixing reinitialization effects with a second adaptation run. The default TAPT settings match the earlier seed-42 comparison.

```python
TAPT_OUT = "artifacts/runs/tapt-muril"
TAPT_LOG = "artifacts/logs/reinit_confirmation_tapt.log"
if pathlib.Path(TAPT_OUT).is_dir():
    print("using existing TAPT checkpoint:", TAPT_OUT)
else:
    run([sys.executable, "-u", "-m", "hastika.task_b.tapt",
         "--out", TAPT_OUT], log=TAPT_LOG)
assert pathlib.Path(TAPT_OUT).is_dir(), "TAPT checkpoint was not written"
print("TAPT checkpoint ready:", TAPT_OUT)
```

## 2. Run both variants at seeds 43 and 44

Each command is a separate five-fold run, which lets the analysis below show whether the one-layer result wins at both confirmation seeds rather than hiding them inside one aggregate.

```python
COMMON = [
    "--model", TAPT_OUT,
    "--folds", "5",
    "--epochs", "6",
    "--select", "last",
    "--aux-weight", "0",
]
ARMS = [
    ("b_tapt_reinit2_s43", "2", "43"),
    ("b_tapt_reinit1_s43", "1", "43"),
    ("b_tapt_reinit2_s44", "2", "44"),
    ("b_tapt_reinit1_s44", "1", "44"),
]
for tag, n_layers, seed in ARMS:
    run([sys.executable, "-u", "-m", "hastika.task_b.train",
         "--tag", tag, *COMMON, "--seeds", seed,
         "--reinit-layers", n_layers], log=f"artifacts/logs/{tag}.log")
print("all four confirmation runs completed")
```

## 3. Recompute scores and summarize the confirmation

The trainer logs its OOF score, and this cell independently recomputes macro-F1 from the saved OOF probabilities. It reports each seed, the mean of the two confirmation seeds, and the probability-averaged score across seeds for each reinitialization setting.

```python
import numpy as np
import pandas as pd
from sklearn.metrics import classification_report, f1_score
from hastika.common.preprocessing import dedupe_index

LABELS = ["Gender", "Geo-political", "Others", "Political", "Religion", "Violence"]
train = pd.read_csv("data/raw/multiclass_train.csv")
keep = dedupe_index(train["Comment"].tolist(), train["Hate Category"].tolist(), "task B")
train = train.iloc[keep].reset_index(drop=True)
y = train["Hate Category"].map(LABELS.index).to_numpy()

records = []
probabilities = {}
for tag, n_layers, seed in ARMS:
    probs = np.load(pathlib.Path("artifacts/runs") / tag / "oof_probs.npy")
    assert probs.shape == (len(y), len(LABELS)), f"unexpected OOF shape for {tag}: {probs.shape}"
    probabilities[(int(n_layers), int(seed))] = probs
    pred = probs.argmax(1)
    report = classification_report(y, pred, labels=range(len(LABELS)),
                                  target_names=LABELS, output_dict=True, zero_division=0)
    records.append({"layers": int(n_layers), "seed": int(seed),
                    "macro_f1": f1_score(y, pred, average="macro"),
                    **{f"f1_{label}": report[label]["f1-score"] for label in LABELS}})

by_seed = pd.DataFrame(records).sort_values(["seed", "layers"])
display(by_seed.round(4))

aggregate = []
for n_layers in [2, 1]:
    seed_scores = by_seed.loc[by_seed["layers"] == n_layers, "macro_f1"]
    mean_probs = np.mean([probabilities[(n_layers, seed)] for seed in [43, 44]], axis=0)
    aggregate.append({"layers": n_layers,
                      "mean_seed_macro_f1": seed_scores.mean(),
                      "mean_probability_macro_f1": f1_score(y, mean_probs.argmax(1), average="macro")})
aggregate = pd.DataFrame(aggregate).set_index("layers").loc[[2, 1]]
display(aggregate.round(4))
print(f"\nOne-layer minus two-layer, mean seed scores: {aggregate.loc[1, 'mean_seed_macro_f1'] - aggregate.loc[2, 'mean_seed_macro_f1']:+.4f}")
print(f"One-layer minus two-layer, averaged probabilities: {aggregate.loc[1, 'mean_probability_macro_f1'] - aggregate.loc[2, 'mean_probability_macro_f1']:+.4f}")

print("\nPer-class reports:")
for n_layers in [2, 1]:
    mean_probs = np.mean([probabilities[(n_layers, seed)] for seed in [43, 44]], axis=0)
    print(f"\nreinit layers = {n_layers}")
    print(classification_report(y, mean_probs.argmax(1), labels=range(len(LABELS)),
                                target_names=LABELS, digits=3, zero_division=0))
```

## 4. Preserve the confirmation outputs

Download the copied directory from Kaggle. If one-layer reinitialization wins at both seeds and in the averaged-probability comparison, update the recommended recipe and rerun the final full-data fit with `--reinit-layers 1`.

```python
OUT = pathlib.Path("/kaggle/working/reinit_confirmation_outputs")
OUT.mkdir(parents=True, exist_ok=True)
for tag, _, _ in ARMS:
    src = pathlib.Path("artifacts/runs") / tag
    dst = OUT / tag
    dst.mkdir(parents=True, exist_ok=True)
    for name in ["oof_probs.npy", "test_probs.npy", "predictions.csv"]:
        if (src / name).exists():
            shutil.copy2(src / name, dst / name)
    shutil.copy2(pathlib.Path("artifacts/logs") / f"{tag}.log", OUT / f"{tag}.log")
if pathlib.Path(TAPT_LOG).exists():
    shutil.copy2(TAPT_LOG, OUT / pathlib.Path(TAPT_LOG).name)
by_seed.to_csv(OUT / "reinit_confirmation_by_seed.csv", index=False)
aggregate.to_csv(OUT / "reinit_confirmation_aggregate.csv")
print("saved outputs to", OUT)
```
